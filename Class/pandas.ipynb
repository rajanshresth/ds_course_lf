```python
import pandas as pd 

def inspect_dataset( dataframe: pd.DataFrame) -> None:
    """Production fn"""

    print(f"=== SHAPE: {dataframe.shape[0]} rows, {dataframe.shape[1]} cols ===")

    summary = pd.DataFrame({
        "Dtype": dataframe.dtypes,
        "Null Count": dataframe.isna().sum(),
        "Null %": (dataframe.isna().sum() / len(dataframe)) * 100,
    })
    stats_summary = pd.DataFrame(dataframe.describe())

    print("\n=== DATA PROFILE ===")
    print(summary)
    print("\n\n=== Stats Summary === \n",stats_summary)

inspect_dataset(df)
```

## `df.loc[row_label, col_label]`
    Label-based indexing. Uses explict columns names and index labels.
    Highly readable and standard in production

## `df.iloc[row_index, col_index]`
    Integer position-based indexing (0,1,...N-1). Slicing behaves like standard python list ( upper bound excluded)

```python
# 1. label-based slicing
high_spenders = df.loc[df['monthly_spend'] > 100,["user_id","churn_label"]]

high_spenders
```

```python
# Integer Positional Slicing --> First 10 rows, first 3 columns
first_10_rows = df.iloc[0:10,0:3]

first_10_rows
```

```python
# Compound Boolean Logic ( & = AND, |= OR, ~ = NOT)
## Boolean masking
target_cohort = df.loc[
    (df["monthly_spend"]>50) & (df["churn_label"] == False),
    ["user_id","signup_date"]
]

target_cohort
```

## Remediation Decision Frameworkabs
1. Missingness > 40-50%: Drop column entirely(`df.drop(columns=['col'])`) to avoid synthetic bias
2. Missingness < 5% (MCAR): Drop affected rows(`df.dropna(subset=['col'])`). Minimal impact on statistical distribution
3. Numerical Imputation: Impute missing entries with median (robust to outlier ) or median
4. Categorical Imputation: Impute missing entries with column Mode (most freq value) or 'unknown'

### Reshaping: `pivot_table()` & `crosstab()`
1. `pivot_table():` Multi-dimensional aggregations for numerical data.
2. `crosstab():` Freq counts and cross-tabulations across categorical dimensions

```python
pivot_df = pd.pivot_table(
    df, values="sales_amount",
    index="customer_segment", columns="category",
    aggfunc="mean", 
    fill_value=0.0, margins=True
)
```

```python
pivot_df
```

```python
freq_table = pd.crosstab(
    index=df["region"],
    columns=df["customer_segment"],
    normalize="index" # Normalize by dividing all values by the sum of values.
)
```

```python
freq_table
```

### Unpivoting Data with melt()

```python
wide_df = pd.DataFrame({
    "Product": ["A","B"],
    "2024": [100, 300],
    "2025": [150,350]
})
wide_df
```

```python
long_df = pd.melt(
    wide_df, 
    id_vars=["Product"],
    value_vars=["2024","2025"],
    var_name="Year",
    value_name="Sales"
)
```
