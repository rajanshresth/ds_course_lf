## Task 1.1: Environmental Sensor Analytics

```python
# A sensor records hourly temperatures over 72 hours. 
# Some readings are corrupted and contain unrealistic values.

np.random.seed(42)

raw_temperatures = np.random.normal(loc=24.0, scale=3.0, size=72)

# Simulated corrupt readings
raw_temperatures[[5, 12, 28, 45, 60]] = [99.9, -15.0, 88.8, -50.0, 102.4]
```

```python
raw_temperatures
```

```python
# 1. Create a boolean mask for valid temperatures.
#    Valid range:
#    15.0 ≤ Temperature ≤ 40.0


valid_mask = (raw_temperatures >= 15) & (raw_temperatures <=40)
valid_temp = raw_temperatures[valid_mask]
```

```python
valid_temp
```

```python
# 1.1.2 Replace all invalid values with the median of the valid temperatures.

valid_temp_median = np.median(raw_temperatures[valid_mask])
print(f"Median of the valid temperatures: {valid_temp_median:.2f}")

raw_temperatures[valid_mask] = valid_temp_median
raw_temperatures
```

```python
# 1.1.3 Calculate and print:
#             Mean
#             Standard Deviation
#             Variance

## Without masking
print("Without Masking:")
unmasked_mean_temp = np.mean(raw_temperatures)
print(f"\tMean Temperature: {unmasked_mean_temp:.2f}")

unmasked_std_temp = np.std(raw_temperatures)
print(f"\tTemperature Standard Deviation: {unmasked_std_temp:.2f}")

unmasked_variance_temp = np.var(raw_temperatures)
print(f"\tTemperature Variance: {unmasked_variance_temp:.2f}")


## With masking
print("With Masking:")
masked_mean_temp = np.mean(raw_temperatures[valid_mask])
print(f"\tMean Temperature: {masked_mean_temp:.2f}")

masked_std_temp = np.std(raw_temperatures[valid_mask])
print(f"\tTemperature Standard Deviation: {masked_std_temp:.2f}")

masked_variance_temp = np.var(raw_temperatures[valid_mask])
print(f"\tTemperature Variance: {masked_variance_temp:.2f}")
```

# Part 2: Sliding Median Filter

```python
# A noisy signal contains random spikes.
signal = np.array([10, 11, 99, 12, 13, -45, 14, 15, 16])
```

```python
filtered_signal = signal.copy()
```

```python
filtered_signal
```

```python
for i in range(1, 8):
    window = signal[i-1 : i+2]
    filtered_signal[i] = np.median(window)
    
print("Original signal:", signal)
print("Filtered signal:", filtered_signal)
```

### Sliding Median Filter Explanation

Step-by-Step Array Updates
* **Loop 1 (`i=1`):** Window is `[10, 11, 99]`. Median is **11**. 
  * Array: `[10, 11, 99, 12, 13, -45, 14, 15, 16]` *(No change)*
* **Loop 2 (`i=2`):** Window is `[11, 99, 12]`. Median is **12**. 
  * Array: `[10, 11, 12, 12, 13, -45, 14, 15, 16]` *(99 is removed)*
* **Loop 3 (`i=3`):** Window is `[99, 12, 13]`. Median is **13**. 
  * Array: `[10, 11, 12, 13, 13, -45, 14, 15, 16]`
* **Loop 4 (`i=4`):** Window is `[12, 13, -45]`. Median is **13**. 
  * Array: `[10, 11, 12, 13, 13, -45, 14, 15, 16]`
* **Loop 5 (`i=5`):** Window is `[13, -45, 14]`. Median is **14**. 
  * Array: `[10, 11, 12, 13, 13, 14, 14, 15, 16]` *(-45 is removed)*

---
The median filter performs significantly better than a mean (average) filter because it completely ignores the actual size of extreme spikes (outliers).
