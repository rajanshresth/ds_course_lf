### Broadcasting Rules
 `Rule: Dimensions are compatible if they are equal, or if one of them is 1`

```python
A = np.ones((3,3))
B = np.array([1,2,3])
C = A + B
C
```

```python
D = np.array([1,2,3]).reshape(3,1)
print("D:\n",D)

E = np.array([1,2,3])
F = D + E
print("\n\nF\n",F)
```

### Boolean Masking
 The most important tool for data cleaning and outlier detection
1. Filter Values based on condintions
2. Replace noise with specific values
3. Combine masks with bitwise operator

```python
data =  np.random.randn(100)
print("data",data)
```

```python
mask = data > 0
print("\n mask",mask)
```

```python
pos_data = data[mask]
pos_data
```

```python
# Remove Outliers
data[data > 0.5] = 0.5
data
```

```python
data[data < -0.5] = 0.5
data
```

```python
# Advanced array shape shifing
arr = np.array([1,2,3,4,5])

# 1. Reshape
print("reshape\n",arr.reshape(5,1))

# 2. Transpose
print("\ntranspose",arr.T)

# 3. Concatenate: Stacking datasets vertically or horizontally
print("\nv_stack\n",np.vstack((1,2,3)))
```

```python
# Lab: Matrix Algebra

A = np.array([[1,2],[3,4]])
B = np.linalg.inv(A)
print("B\n", B)

I = A @ B
print("\nMatrix Mult: (I = A @ B) \n", I)

eig_vals = np.linalg.eigvals(A)
print("\nEigen Values:\n",eig_vals)
```
